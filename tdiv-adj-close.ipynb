```python
import yfinance as yf
import pandas as pd

ticker = yf.Ticker("TDIV.AS")
history = ticker.history(period="2y", interval="1d")
```

```python

h = len(history)//2
m1 = history.Close[:h].mean()
m2 = history.Close[h:].mean()
print(f"Mean price for the first year: {m1:.2f}")
print(f"Mean price for the second year: {m2:.2f}")
print(f"Price change: {m2 / m1 - 1:.2%}")
```

```python
dl = yf.download("TDIV.AS", period="2y", interval="1d", auto_adjust=False)
```

```python
h = len(dl)//2
m1 = dl["Close","TDIV.AS"][:h].mean()
m2 = dl["Close","TDIV.AS"][h:].mean()
change = m2 / m1 - 1
print(f"Mean price for the first year: {m1:.2f}")
print(f"Mean price for the second year: {m2:.2f}")
print(f"Price change: {change:.2%}")

adj_change = dl["Adj Close","TDIV.AS"][h:].mean() / dl["Adj Close","TDIV.AS"][:h].mean() - 1
print(f"Adjusted price change: {adj_change:.2%}")
```

```python
from yfetch import get_stock_history
```

```python
symbol = 'SPMO'

history = get_stock_history(symbol, interval='1d')
print(f'{symbol} history: {len(history)}d')
```

```python
print(' Cal   Tr')
print('days days   -> SMA      now')
for w in range(1, 33):
    cal_days = w * 7
    tr_days = w * 5
    if tr_days > len(history):
        break

    sma = history['Close'].rolling(window=tr_days).mean()
    sma_change = (sma / history['Close'] - 1).median()
    last = sma.iloc[-1] / history['Close'].iloc[-1] - 1
    print(f'{cal_days:4} {tr_days:4} {sma_change:+8.2%} {last:+8.2%}')
```

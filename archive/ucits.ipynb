```python
import yfinance as yf
from datetime import datetime
import pandas as pd
from time import sleep


data_file = 'data/ucits.csv'
df = pd.read_csv(data_file)
rows = []
for _, row in df.iterrows():
    symbol = row['Symbol']
    revolut = row['Revolut']
    sleep(1)
    ticker = yf.Ticker(symbol)
    info = ticker.info
    name = info.get('longName') or info.get('shortName')
    print(f'{symbol} {name}')
    rows.append({
        'Symbol': symbol,
        'Net Assets (MM)': round(info['netAssets'] / 1_000_000) if info.get('netAssets') else None,
        'Inception Date': info.get('fundInceptionDate') and 
            datetime.fromtimestamp(info['fundInceptionDate']).date().isoformat(),
        'Revolut': revolut,
        'Name': name,
    })

df_output = pd.DataFrame(rows).sort_values(by='Symbol')
df_output.to_csv(data_file, index=False)
display(df_output)
```

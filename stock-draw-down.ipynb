```python

from yfetch import get_stock_history


symbol = 'NVDA'

history = get_stock_history(symbol, interval='1wk')
# display(history)
```

```python
window = round(144/7)

# Get the open price at the start of each rolling window
open_at_start = history['Open'].shift(window - 1)

# Get the minimum low price in the rolling window
min_low = history['Low'].rolling(window=window).min()

# Calculate drawdown as percentage (0-1)
history['Drawdown'] = min_low / open_at_start - 1

print(f"{symbol} - {window} Week Drawdown median: {history['Drawdown'].median():.2%}")

display(history)
```

```python
import plotly.express as px

fig = px.histogram(history, x='Drawdown', nbins=30, title=f'{symbol} - {window} Week Drawdown Distribution')
fig.show()
```
